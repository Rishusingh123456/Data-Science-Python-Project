```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("diabetes_cleaned.csv")
```

```python
print(df['Diabetes_012'].value_counts())
```

```python
print(df['Diabetes_012'].value_counts(normalize=True)*100)
```

```python
plt.figure(figsize=(7,5))
sns.countplot(x="Diabetes_012",data=df)
plt.title("Distribution of Diabetes Status")
plt.xlabel("Diabetes Status")
plt.ylabel("Number of Individuals")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["BMI"],kde=True)
plt.title("BMI Distribution")
plt.xlabel("BMI")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="Diabetes_012",y="BMI",data=df)
plt.title("BMI Distribution by Diabetes Status")
plt.xlabel("Diabetes Status")
plt.ylabel("BMI")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x="HighBP",hue="Diabetes_012",data=df)
plt.title("Diabetes Status by High Blood Pressure")
plt.xlabel("High Blood Pressure")
plt.ylabel("Number of Individuals")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x="HighChol",hue="Diabetes_012",data=df)
plt.title("Diabetes Status by High Cholesterol")
plt.xlabel("High Cholesterol")
plt.ylabel("Number of Individuals")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x="PhysActivity",hue="Diabetes_012",data=df)
plt.title("Diabetes Status by Physical Activity")
plt.xlabel("Physical Activity")
plt.ylabel("Number of Individuals")
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.countplot(x="GenHlth",hue="Diabetes_012",data=df)
plt.title("General Health by Diabetes Status")
plt.xlabel("General Health")
plt.ylabel("Number of Individuals")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x="Age",hue="Diabetes_012",data=df)
plt.title("Diabetes Status Across Age Categories")
plt.xlabel("Age Category")
plt.ylabel("Number of Individuals")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.countplot(x="Sex",hue="Diabetes_012",data=df)
plt.title("Diabetes Status by Sex")
plt.xlabel("Sex")
plt.ylabel("Number of Individuals")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.countplot(x="Smoker",hue="Diabetes_012",data=df)
plt.title("Diabetes Status by Smoking Status")
plt.xlabel("Smoker")
plt.ylabel("Number of Individuals")
plt.show()
```

```python
corr = df.corr(numeric_only=True)
```

```python
plt.figure(figsize=(16,12))
sns.heatmap(corr,annot=True,fmt=".2f")
plt.title("Correlation Matrix of Diabetes Health Indicators")
plt.show()
```

```python
summary = df.groupby("Diabetes_012")[[
    "BMI",
    "HighBP",
    "HighChol",
    "GenHlth",
    "MentHlth",
    "PhysHlth",
    "Age",
    "Income"
]].mean()
print(summary)
```
