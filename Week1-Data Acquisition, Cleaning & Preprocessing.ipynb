```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("diabetes_012_health_indicators_BRFSS2015.csv")
df.head()
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
df.columns
```

```python
df.info()
```

```python
df.describe()
```

```python
missing = df.isnull().sum()
print(missing)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100
print(missing_percentage)
```

```python
duplicate_count = df.duplicated().sum()
print("Duplicate rows:", duplicate_count)
```

```python
df_clean = df.drop_duplicates().copy()
print(df_clean.shape)
```

```python
print("Rows before:", len(df))
print("Rows after:", len(df_clean))
print("Rows removed:", len(df) - len(df_clean))
```

```python
for column in df_clean.columns:
    print("\n", column)
    print(df_clean[column].unique())
```

```python
print(df_clean["Diabetes_012"].value_counts())
```

```python
print(df_clean["Diabetes_012"].value_counts(normalize=True) * 100)
```

```python
sns.countplot(
    x="Diabetes_012",
    data=df_clean
)

plt.title("Distribution of Diabetes Status")
plt.xlabel("Diabetes Status")
plt.ylabel("Number of Individuals")
plt.show()
```

```python
print(df_clean["BMI"].describe())
```

```python
sns.boxplot(x=df_clean["BMI"])

plt.title("BMI Outlier Analysis")
plt.xlabel("BMI")
plt.show()
```

```python
numerical_columns = [
    "BMI",
    "MentHlth",
    "PhysHlth"
]

for column in numerical_columns:
    sns.boxplot(x=df_clean[column])
    plt.title(f"{column} Outlier Analysis")
    plt.show()
```
